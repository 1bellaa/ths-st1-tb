```python
import subprocess
from pathlib import Path
from tqdm import tqdm
```

```python
def bcftools_filter(sample, input_vcf, output_vcf, log_file_path):
    if output_vcf.exists():
        print(f"⏭️ Skipping {sample}: SNP VCF already exists.")
        return

    if not input_vcf.exists():
        print(f"❌ Missing input VCF: {input_vcf}")
        return

    # bcf_cmd = [
    #     "bash", "-c",
    #     f"bcftools view -v snps {input_vcf} | "
    #     f"bcftools filter -i 'INFO/DP>=5 && QUAL>=20 && AF>=0.75' "
    #     f"-o {output_vcf} --output-type v"
    # ]

    bcf_cmd = [
        "bcftools", "view",
        "-v", "snps",
        "-i", "INFO/DP>=5 && QUAL>=20 && AF>=0.75",
        input_vcf,
        "-o", str(output_vcf),
        "--output-type", "v"
    ] 

    try:
        with open(log_file_path, "w") as log_file:
            subprocess.run(bcf_cmd, stdout=log_file, stderr=log_file, check=True)
        print(f"✅ BCFtools done for {sample}")
    except subprocess.CalledProcessError as e:
        print(f"❌ BCFtools failed for {sample}: {e}")
```

```python
# === Batch Runner ===
vcf_dir = Path("./vcf")
snp_only_dir = Path("./snp_only")
logs_dir = Path("./logs")

vcf_files = list(vcf_dir.glob("*.targets.vcf.gz"))

print(f"📁 Found {len(vcf_files)} VCF files to process.")
```

```python
for vcf_file in tqdm(vcf_files, desc="🧬 Filtering SNPs"):
    sample = vcf_file.stem.replace(".targets", "")
    output_vcf = snp_only_dir / f"{sample}_snps_only.vcf"
    log_file = logs_dir / f"{sample}_bcftools.log"
    bcftools_filter(sample, vcf_file, output_vcf, log_file)
```
